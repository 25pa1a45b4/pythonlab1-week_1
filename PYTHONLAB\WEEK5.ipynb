```python
a=str(input())
b=a.lower()
vowels="aeiou"
countc=0
countv=0
for ch in b:
  if ch in vowels:
    countv+=1
  else:
    countc+=1
print("no of vowels are",countv)
print("no of cons are",countc)
```
