```python
# 1. Creating a set
s = {1, 2, 3, "hello", 1, 2, 3}
print(s)
```

```python
# 2. Empty set
A = {}
print(type(A))
```

```python
# 3. Union, intersection, difference, symmetric difference
A = {1, 2, 3}
B = {3, 4, 5}

print(A | B)         # Union
print(A & B)         # Intersection
print(A - B)         # Difference
print(A ^ B)         # Symmetric difference
```

```python
# 4. Set from lists
L = [12, 33, 44]
L1 = [33, 45, 67]

A = set(L)
B = set(L1)

print(A & B)
```

```python
# 5. Set with different data types
s = {1, "hello", 22.3, (1, 2, 3)}
print(s)
```

```python
# 6. Adding elements
s.add(23)
print(s)
```

```python
A = {1, 2, 3}
B = {3, 4, 5}
print(A, B)
```

```python

s = set()
s.update((1, 2, 3, 4, 5))

s.add(10)
print(s)

s.remove(10)
print(s)

s.discard(100)
print(s)

s.pop()
print(s)

s.clear()
print(s)
```

```python
n = 5
numbers = set(range(1, n + 1))
L = [1, 1, 2, 2, 5, 3]
L2 = set(L)
print(numbers - L2)
```
