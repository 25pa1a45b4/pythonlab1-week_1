```python
# check wheather a given number is palindrome or not using functions
def palindrome(s):

  s=s.lower()
  return s==s[::-1]
word=input()
if palindrome(word):
  print("correct")
else:
  print("not")
```

```python
# program to reverse a string
def reverse(name):
  return name[::-1]
name=input()
print(reverse(name))
```
