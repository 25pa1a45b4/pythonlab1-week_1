```python
n=int(input("enter a number"))
for i in range(2,n):
  if n%i==0:
    print("not prime")
    break
else:
    print("prime")
```

```python
#print n to m prime numbers
n=int(input("enter a number"))
m=int(input("enter another number"))
for i in range(n,m+1):
  for j in range(2,i):
    if(i%j==0):
      break
  else:
      print(i)
```

```python
n=int(input("enter a number"))
sum=0
while n>0:
  digit=n%10
  sum=sum+digit
  n=n//10
print(f"sum is {sum}")
```
