```python
#OPERATIONS ON METHODS
#data type
tuples1=(10,1,2,-50,70)
tuples2=("akshaya",99,"bittu","rupa","greaty")
tuples3=(10)
tuples4=(8,99,"akku")
tuples5=(9,77,56,90,22)
print(type(tuples1))
print(type(tuples2))
print(type(tuples3))

#indexing
print(tuples1[2])
print(tuples2[1])

#slicing
print(tuples2[1:3])
print(tuples1[:])

#length
print(len(tuples1))
print(len(tuples2))

#concatination
print(tuples1+tuples2)
print(tuples1+tuples4)

#repitation
print(tuples1*2)
print(tuples4*10)

#membership

print(10 not in tuples2)
print("akshaya" in tuples4)


```

```python
#methods
 #count
print(tuples2.count(99))
print(tuples2.count("akshaya"))
#index
print(tuples4.index(99))
print(tuples1.index(-50))
#functions
print(len(tuples1))
print(max(tuples5))
print(min(tuples1))

print(sum(tuples5))
print(tuple(sorted(tuples1)))
```

```python
#zip method
names=("akshaya","jhansi","john")
marks=(99,100,68)
result=zip(names,marks)
print(tuple(result))
```
